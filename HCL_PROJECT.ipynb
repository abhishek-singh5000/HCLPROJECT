```python
import numpy as np
import pandas as pd

# 1. Create the dataset of 10 basic General Knowledge questions using Pandas
reply = input("Enter your name :")
print( "Let's start the quiz",reply)

quiz_data = {
    "Question": [
        "What is the capital city of India?",
        "Which planet is known as the Red Planet?",
        "Who wrote the Indian National Anthem 'Jana Gana Mana'?",
        "What is the largest ocean on Earth?",
        "Which gas do plants absorb from the atmosphere?",
        "How many continents are there on Earth?",
        "What is the chemical symbol for Gold?",
        "Which is the longest river in the world?",
        "Who is known as the Father of the Indian Constitution?",
        "In which year did India gain independence?"
    ],
    "Option_A": ["Mumbai", "Venus", "Bankim Chandra Chatterjee", "Atlantic Ocean", "Oxygen", "5", "Go", "Amazon", "Mahatma Gandhi", "1945"],
    "Option_B": ["New Delhi", "Mars", "Rabindranath Tagore", "Indian Ocean", "Carbon Dioxide", "6", "Gd", "Nile", "Jawaharlal Nehru", "1947"],
    "Option_C": ["Kolkata", "Jupiter", "Sarojini Naidu", "Arctic Ocean", "Nitrogen", "7", "Au", "Ganga", "Dr. B.R. Ambedkar", "1950"],
    "Option_D": ["Chennai", "Saturn", "Subhas Chandra Bose", "Pacific Ocean", "Hydrogen", "8", "Ag", "Yangtze", "Sardar Patel", "1952"],
    "Answer": ["B", "B", "B", "D", "B", "C", "C", "B", "C", "B"]
}

# Load into a Pandas DataFrame and shuffle the questions randomly
df = pd.DataFrame(quiz_data)
df = df.sample(frac=1).reset_index(drop=True)

# 2. Initialize a NumPy array to track results (1 for correct, 0 for incorrect)
scores = np.zeros(len(df), dtype=int)

print("=== WELCOME TO THE GENERAL KNOWLEDGE QUIZ ===\n")

# 3. Run the quiz loop
for index, row in df.iterrows():
    print(f"Q{index + 1}: {row['Question']}")
    print(f"  A) {row['Option_A']}")
    print(f"  B) {row['Option_B']}")
    print(f"  C) {row['Option_C']}")
    print(f"  D) {row['Option_D']}")

    user_ans = input("Enter your choice (A/B/C/D): ").strip().upper()

    if user_ans == row['Answer']:
        print("Correct!\n")
        scores[index] = 1
    else:
        print(f"Wrong! The correct answer was {row['Answer']}.\n")

# 4. Use NumPy to calculate final statistics
total_score = np.sum(scores)
accuracy = np.mean(scores) * 100

# Use np.where for simple conditional grading
grade = np.where(accuracy >= 80, "Excellent", np.where(accuracy >= 50, "Good", "Needs Practice"))

# 5. Summary Report using Pandas
df['Your_Score'] = scores
print("=== QUIZ SUMMARY ===")
print(f"Total Score : {total_score} / {len(df)}")
print(f"Accuracy    : {accuracy:.1f}%")
print(f"Performance : {grade}")
```
